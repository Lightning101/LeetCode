***This problem can be solved using a stack.***

* The approach is to put all the elements onto the stack and when a opperator is found pop the top 2 elements and perform the transformation.

### Complexity:

***Time***:O(n)

***Space***: O(n)

```python
class Solution:

    def evalRPN(self, tokens) -> int:

        st = []
        ops = {
            "+": lambda n1, n2: n1 + n2,
            "-": lambda n1, n2: n1 - n2,
            "*": lambda n1, n2: n1 * n2,
            "/": lambda n1, n2: int(n1 / n2),
        }

        ans = 0
        for token in tokens:

            if token in ops:
                n2 = st.pop()
                n1 = st.pop()
                st.append(ops[token](n1, n2))
            else:
                st.append(int(token))

        return st[0]
```
