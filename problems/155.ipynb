### Comments
***This problem can be solved using a monotonic stack***


* The key lies in creating 2 lists and using one as a monotonic stack.
* This way we have fast access to the smallest element at anytime.

!Note
* Watch out for duplicate elements :).

### Complexity:

***Time***:O(1)

* All opperations should be ~O(1)

***Space***:O(n)

* Where n is the number of elements add to the list.

````python
class MinStack:

    def __init__(self):
        self.data = []
        self.min = []

    def push(self, val: int) -> None:
        if not self.min or self.min[-1] >= val:
            self.min.append(val)

        self.data.append(val)

    def pop(self) -> None:
        if self.data[-1] == self.min[-1]:
            self.min.pop()
        return self.data.pop()

    def top(self) -> int:
        return self.data[-1]

    def getMin(self) -> int:
        return self.min[-1]
````

### Community References:


* Alternate approach that only uses one stack

* By keeping the min element at the time we get better space usage.

```
class MinStack:

    def __init__(self):
        self.stack = []

    def push(self, x: int) -> None:

        # If the stack is empty, then the min value
        # must just be the first value we add
        if not self.stack:
            self.stack.append((x, x))
            return

        current_min = self.stack[-1][1]
        self.stack.append((x, min(x, current_min)))

    def pop(self) -> None:
        self.stack.pop()

    def top(self) -> int:
        return self.stack[-1][0]

    def getMin(self) -> int:
        return self.stack[-1][1]
```
