* The best aproach I could come up with in 15min time slot was to attack the n! factorial from both ends. Which would reduce our time when input bounded from 10^4 -> 5^4.

### Complexity:

***Time***:O(n^2)

***Space***: O(nlogn)

```python
import math


class Solution:
    def trailingZeroes(self, n: int) -> int:
        if n == 0:
            return n
        val = 1

        ans = 0

        for i in range(1, (n + 2) // 2):
            val *= i * (n + 1 - i)

        if n % 2:
            val *= math.ceil(n / 2)

        while val % 10 == 0:
            ans += 1
            val //= 10

        return ans
```
