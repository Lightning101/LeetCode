### Comments
***This problem can be solved using two pointer apporach***

* The key lies in intializing 2 pointers 1 fast and 1 slow.
* We iterate the fast pointer by n and continue in a linear fashion.

* Edge cases are when the head itself needs to be removed.

### Complexity:

***Time***: O(n)

***Space***: O(1)

```python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def removeNthFromEnd(self, head: Optional[ListNode], n: int) -> Optional[ListNode]:

        slow, fast = head, head

        for i in range(n):
            fast = fast.next

        prev = None

        while fast:
            prev = slow
            slow = slow.next
            fast = fast.next

        if prev:
            prev.next = slow.next if (slow) else None

        else:
            head = head.next
        return head
```
