```
import heapq


class Solution:
    def kSmallestPairs(
        self, nums1: List[int], nums2: List[int], k: int
    ) -> List[List[int]]:
        h = []

        for i in range(len(nums1)):
            for j in range(len(nums2)):
                if len(h) < k:
                    heapq.heappush(h, [-(nums1[i] + nums2[j]), nums1[i], nums2[j]])
                else:
                    heapq.heappushpop(h, [-(nums1[i] + nums2[j]), nums1[i], nums2[j]])

        return [heapq.heappop(h)[1:] for i in range(k)]
```

* The answer is based of a community solution.

* The idea is the same process used in Dijkstra algorithm. This allows us to take the smallest pair optimizing at each step.


### Complexity:

***Time***: O(min(k,m*n))

***Space***: O(min(k,m*n))

````python
import heapq


class Solution:
    def kSmallestPairs(
        self, nums1: List[int], nums2: List[int], k: int
    ) -> List[List[int]]:
        ans = []
        seen = set()
        h = [(nums1[0] + nums2[0], (0, 0))]
        while k > 0 and h:
            val, (i, j) = heapq.heappop(h)
            ans.append([nums1[i], nums2[j]])

            if i + 1 < len(nums1) and not (i + 1, j) in seen:
                heapq.heappush(h, (nums1[i + 1] + nums2[j], (i + 1, j)))
                seen.add((i + 1, j))
            if j + 1 < len(nums2) and not (i, j + 1) in seen:
                heapq.heappush(h, (nums1[i] + nums2[j + 1], (i, j + 1)))
                seen.add((i, j + 1))
            k -= 1

        return ans
````
