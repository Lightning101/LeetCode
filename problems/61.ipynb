* The first steps are taken from rotate an array algorithem.
 * Take the size get k = k % size to give us how many rotations needed.
 * then reverse the array and swap the first and last k elements.

* Find the size calculate the new k.
* Then slow pointer fast pointer approach to find the rotation point.
* Then swap the pointers

### Complexity:

***Time***: O(n)

***Space***: O(1)

```python
from collections import deque


# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def rotateRight(self, head: Optional[ListNode], k: int) -> Optional[ListNode]:

        if not head:
            return None
        if not head.next or k == 0:
            return head

        size = 0

        node = head

        while node:
            size += 1
            node = node.next

     
        k = k % size

        if k == 0:
            return head

        slow = head
        fast = head

        for _ in range(k):
            fast = fast.next

        while fast.next:
            slow = slow.next
            fast = fast.next

        ans = slow.next
        slow.next = None
        fast.next = head

        return ans
```
