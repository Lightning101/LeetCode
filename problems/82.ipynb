### Comments
***This problem can be solved using a set***

* The key to solving this problem lies in using a set and forward iterating the linked list.
* This way we keep values we have already encountered in the set and us it to filter out nodes in our answer list.

### Complexity:

***Time***: O(n)

***Space***: O(n)

````python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def deleteDuplicates(self, head: Optional[ListNode]) -> Optional[ListNode]:
        seen = set()

        ans = ListNode()
        dummy = ans

        node = head
        while node:
            seen.add(node.val)
            curr = node
            node = node.next
            while node and node.val in seen:
                node = node.next
                curr = None
            if curr:
                dummy.next = curr
                dummy = dummy.next
                dummy.next = None
        return ans.next
````
